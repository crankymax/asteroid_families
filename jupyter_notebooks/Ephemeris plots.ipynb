# Family Ephemeris plots

These plots use the ephemeris data from Horizons to check for outliers and anomolies.


```python
import pandas as pd
import numpy as np
import h5py as h5
from pathlib import Path

import datetime as datetime
import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
family_name = "euterpe"

project_dir = Path(f'../../{family_name}_ast/')
plot_dir = project_dir / 'plots'

family = pd.read_hdf(project_dir /'ephemerides.h5')
```

```python
plt.scatter(data=family, x='X',y='Y', marker=".")
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.xlabel('X [AU]')
plt.ylabel('Y [AU]')
plt.show()
```

```python
plt.scatter(data=family, x='X',y='Z', marker=".")
plt.ylim(-0.25, 0.25)
plt.xlim(-4, 4)
plt.xlabel('X [AU]')
plt.ylabel('Z [AU]')

plt.show()
```

```python
plt.scatter(data=family, x='Y',y='Z', marker=".")
plt.ylim(-0.25, 0.25)
plt.xlim(-4, 4)
plt.xlabel('Y [AU]')
plt.ylabel('Z [AU]')

plt.show()
```

```python
# This can be used to identify extreme asteroids

family[(family.Y > 2.7) | (np.abs(family.Z) > 0.8)]
```
