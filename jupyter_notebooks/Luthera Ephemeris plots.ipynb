# Luthera Ephemeris plots
Plots of the ephemeris data 


```python
import pandas as pd
import numpy as np
import h5py as h5

import datetime as datetime
import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
luthera = pd.read_hdf('../asteroid_data/luthera_eph.h5')
```

```python
plt.scatter(data=luthera, x='X',y='Y', marker=".")
plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.xlabel('X [AU]')
plt.ylabel('Y [AU]')
plt.show()
```

```python
plt.scatter(data=luthera, x='X',y='Z', marker=".")
plt.ylim(-2, 2)
plt.xlim(-5, 5)
plt.xlabel('X [AU]')
plt.ylabel('Z [AU]')

plt.show()
```

```python
plt.scatter(data=luthera, x='Y',y='Z', marker=".")
plt.ylim(-2, 2)
plt.xlim(-5, 5)
plt.xlabel('Y [AU]')
plt.ylabel('Z [AU]')

plt.show()
```
