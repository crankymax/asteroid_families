# Luthera Family Plots



```python
import pandas as pd
import numpy as np
import h5py as h5

import datetime as datetime

import matplotlib.pyplot as plt
```

```python
path = './luthera_family.h5'
luth_fam = pd.read_hdf(path)
```

```python
luth_fam['ideg'] = np.degrees(np.arcsin(luth_fam['sineI_P']))
```

```python
luth_fam.describe()
```

```python
plt.scatter(data=luth_fam, x='a_P',y='e_P', edgecolors='black', marker=".")
plt.xlabel("Proper semi-major axis $a_P$ [AU]")
plt.ylabel("Proper eccentricity $e_P$ []")
plt.title("Proper semi-major axis vs proper eccentricity")
plt.show()
```

```python
plt.scatter(data=luth_fam, x='a_P',y='ideg', edgecolors='black', marker=".")
plt.xlabel("Proper semi-major axis $a_P$ [AU]")
plt.ylabel("Proper inclination $i$ [$\degree$]")
plt.title("Proper semi-major axis vs $\sin i_P$")
plt.show()
```

```python
plt.scatter(data=luth_fam, x='a_P',y='abs_mag' , edgecolors='black', marker=".")
plt.xlabel("Proper semi-major axis $a_P$ [AU]")
plt.ylabel("Absolute $H / mag$ []")
plt.title("Proper semi-major axis vs $H / mag$")
plt.show()
```

```python
plt.hist(data=luth_fam, x='abs_mag')
plt.ylabel("Count")
plt.xlabel("Absolute $H / mag$ []")
plt.title("$H / mag$ histogram")
plt.show()
```

```python
plt.hist(data=luth_fam, x='a_P', bins=40)
plt.ylabel("Count")
plt.xlabel("$a_P$ [AU]")
plt.title("$a_P$ histogram")
plt.show()
```
